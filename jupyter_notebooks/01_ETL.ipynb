# ETL: Extraction

We will be working with numbers and arrays and therefore the Numpy library is extremely useful in Python. To load a CSV file with Python, the Pandas library is essential. It has a specific function that reads CSV files by parsing the file path. We will import both libraries in the Jupyter Notebook.

```python
import pandas as pd
import numpy as np
```

```python
import glob
# Get all CSV file paths in the folder
csv_files = glob.glob("data/raw/*.csv")

print(f"Found {len(csv_files)} files:")
for f in csv_files:
    print(f)
```

```python
df_list = []

for file in csv_files:
    # Read each CSV file
    temp_df = pd.read_csv(file)
    
    # Extract the year from the filename (e.g., "FIFA - 1930.csv" → 1930)
    year = os.path.basename(file).split(" - ")[-1].split(".")[0]
    temp_df["Year"] = year  # Add a column for World Cup year
    
    # Add to list
    df_list.append(temp_df)

# Combine all DataFrames into one
combined_df = pd.concat(df_list, ignore_index=True)
```

```python
print(combined_df.shape)
combined_df.head()
```
